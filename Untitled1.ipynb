```python
import tsl
import torch
import numpy as np
import pandas as pd
from tsl.datasets import MetrLA
from torch_geometric.utils.undirected import is_undirected
from torch_geometric.utils.loop import remove_self_loops
from torch_geometric.utils.isolated import contains_isolated_nodes
from tsl.data import SpatioTemporalDataset
from tsl.data.datamodule import (SpatioTemporalDataModule,
                                 TemporalSplitter)
from tsl.data.preprocessing import StandardScaler

import torch
from torch_cluster import random_walk
```

```python
dataset = MetrLA(root='./data')
```

```python
connectivity = dataset.get_connectivity(threshold=0.4,
                                        include_self=False,
                                        # normalize_axis=1,
                                        force_symmetric=True,
                                        layout="edge_index")
```

```python
connectivity[1]
```

```python
connectivity[1].T.shape
```

```python
torch_dataset = SpatioTemporalDataset(target=dataset.dataframe(),
                                      connectivity=connectivity,
                                      mask=dataset.mask,
                                      horizon=12,
                                      window=12,
                                      stride=1)
print(torch_dataset)
```

```python
# Normalize data using mean and std computed over time and node dimensions
scalers = {'target': StandardScaler(axis=(0, 1))}

# Split data sequentially:
#   |------------ dataset -----------|
#   |--- train ---|- val -|-- test --|
splitter = TemporalSplitter(val_len=0.1, test_len=0.2)

dm = SpatioTemporalDataModule(
    dataset=torch_dataset,
    scalers=scalers,
    splitter=splitter,
    batch_size=64,
)

dm.setup()
print(dm)
```
